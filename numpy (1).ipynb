Importar numpy

```python
import numpy as np
```

Criar o array de vendas

```python
pecas_vendidas = np.array([150, 120, 90, 210, 300, 250, 180])

print(pecas_vendidas)
```

Calcular o faturamento

```python
faturamento = pecas_vendidas * 50

print(faturamento)
```

Filtrar dias de pico

```python
mascara = pecas_vendidas > 200

print(mascara)
print(pecas_vendidas[mascara])
```

Criar a matriz

```python
vendas = np.array([
    [200, 220, np.nan, 250],
    [150, 180, 160, 190],
    [300, 310, 290, 330]
])

print(vendas)
```

Total por loja com axis=1 (soma horizontal)

```python
total_lojas = np.sum(vendas, axis=1)

print(total_lojas)
```

A soma da Loja A resulta em nan porque existe uma falha no terceiro dia. Para calcular o total ignorando esse valor, use:

```python
total_lojas = np.nansum(vendas, axis=1)

print(total_lojas)
```

Média diária geral ignorando NaN

```python
media_rede = np.nanmean(vendas)

print(media_rede)
```

Verificar a meta diária

```python
resultado_meta = np.where(
    vendas >= 200,
    'Meta Atingida',
    'Abaixo'
)

print(resultado_meta)
```

Gerar 50 categorias aleatórias

```python
rng = np.random.default_rng(42)

categorias = np.array(['Eletrônicos', 'Roupas', 'Casa'])

vendas_categorias = rng.choice(categorias, size=50)

print(vendas_categorias)
```

Contar vendas por departamento

```python
departamentos, contagens = np.unique(
    vendas_categorias,
    return_counts=True
)

print(departamentos)
print(contagens)
```

Encontrar a campanha campeã

```python
faturamento_campanhas = np.array([12000, 45000, 23000, 89000, 31000])

indice_campea = np.argmax(faturamento_campanhas)

print(indice_campea)
print(faturamento_campanhas[indice_campea])
```

Organização e transformação de dados - Dias do mês

```python
dias = np.arange(1, 31)

print(dias)
```

Metas igualmente espaçadas

```python
metas = np.linspace(20000, 30000, 5)

print(metas)
```

Converter decimais para inteiros

```python
estoque = np.array([10.5, 20.1, 30.9])

estoque_inteiro = estoque.astype(int)

print(estoque_inteiro)
```

Últimos cinco dias em ordem invertida

```python
ultimos_dias = dias[-5:][::-1]

print(ultimos_dias)
```

Redimensionamento e proteção de dados - Reshape de 12 meses

```python
visitas = np.array([100, 120, 140, 160, 180, 200, 220, 240, 260, 280, 300, 320])

visitas_trimestres = visitas.reshape(4, 3)

print(visitas_trimestres)
```

Empilhar matrizes

```python
primeiro_semestre = np.array([
    [100, 120, 140],
    [160, 180, 200]
])

segundo_semestre = np.array([
    [220, 240, 260],
    [280, 300, 320]
])

ano = np.vstack((primeiro_semestre, segundo_semestre))

print(ano)
```

Copiar o primeiro trimestre

```python
projecao = visitas_trimestres[0, :].copy()

projecao[0] = 999

print(projecao)
print(visitas_trimestres[0])
```

Achatar a matriz

```python
visitas_1d = ano.ravel()

print(visitas_1d)
```

Ranking com argsort()

```python
pontuacoes = np.array([85, 92, 78, 95, 88])

ranking = np.argsort(pontuacoes)

print(ranking)
```

ranking do maior para o menor:

```python
ranking_desc = np.argsort(pontuacoes)[::-1]

print(ranking_desc)
```

Broadcasting com np.newaxis

```python
precos = np.array([100, 200, 300])

precos_coluna = precos[:, np.newaxis]

print(precos_coluna)
print(precos_coluna.shape)
```

aplicar três descontos simultaneamente:

```python
descontos = np.array([0.10, 0.20, 0.30])

precos_com_descontos = precos_coluna * (1 - descontos)

print(precos_com_descontos)
```

Resolver sistema de custos com np.linalg.solve()

```python
A = np.array([
    [2, 1],
    [1, -1]
])

b = np.array([500, 100])

custos = np.linalg.solve(A, b)

print(custos)
```
